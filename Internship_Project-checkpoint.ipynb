```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load Dataset
df = pd.read_csv("Dataset .csv")
df
```

```python
# Display first 5 rows
print(df.head())

# Dataset Information
print(df.info())

# Check Missing Values
print(df.isnull().sum())
```

## LEVEL 1


### Task 1: Top Cuisines
### Top 3 Most Common Cuisines

```python
# Split cuisines and count frequency

cuisine_count = df['Cuisines'].dropna().str.split(', ').explode()

top_cuisines = cuisine_count.value_counts().head(3)

print("Top 3 Cuisines:")
print(top_cuisines)
```

### Percentage of Restaurants Serving Top Cuisines

```python
total_restaurants = len(df)

for cuisine in top_cuisines.index:
    count = df['Cuisines'].fillna('').str.contains(cuisine).sum()
    percentage = (count / total_restaurants) * 100

    print(f"{cuisine}: {percentage:.2f}%")
```

### Visualization

```python
plt.figure(figsize=(8,5))
top_cuisines.plot(kind='bar')

plt.title('Top 3 Cuisines')
plt.xlabel('Cuisine')
plt.ylabel('Count')
plt.show()
```

### Task 2: City Analysis
### City with Highest Number of Restaurants

```python
city_counts = df['City'].value_counts()

print("City with Highest Number of Restaurants:")
print(city_counts.head(1))
```

### Average Rating by City

```python
city_ratings = df.groupby('City')['Aggregate rating'].mean()

print(city_ratings.sort_values(ascending=False))
```

### Highest Average Rated City

```python
highest_city = city_ratings.idxmax()
highest_rating = city_ratings.max()

print("Highest Rated City:", highest_city)
print("Average Rating:", highest_rating)
```

### Visualization

```python
top10 = city_counts.head(10)

plt.figure(figsize=(10,5))
top10.plot(kind='bar')

plt.title('Top 10 Cities by Restaurant Count')
plt.show()
```

### Task 4: Restaurant Chains
### Identify Chains

```python
chains = df['Restaurant Name'].value_counts()

chains = chains[chains > 1]

print(chains.head(20))
```

### Ratings of Chains

```python
chain_ratings = df.groupby(
    'Restaurant Name'
)['Aggregate rating'].mean()

print(
    chain_ratings.loc[chains.index]
    .sort_values(ascending=False)
    .head(10)
)
```

### Popularity Using Votes

```python
chain_votes = df.groupby(
    'Restaurant Name'
)['Votes'].sum()

print(
    chain_votes.loc[chains.index]
    .sort_values(ascending=False)
    .head(10)
)
```

### Visualization

```python
top_chains = chains.head(10)

plt.figure(figsize=(10,5))

top_chains.plot(kind='bar')

plt.title('Top Restaurant Chains')
plt.ylabel('Number of Branches')

plt.show()
```
